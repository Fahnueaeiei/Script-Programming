# lab4_1.py - Student Management System (List Operations)

```python
# Lab 4.1: Student Management System (List Operations)

print("--- Lab 4.1: Student Management System ---")

# 1) Create an empty list
student_names = []

# 2) Add 3 student names
print("\\n[1] Add Students")
for i in range(3):
    name = input(f"Enter student name #{i + 1}: ")
    student_names.append(name)

print(f"Current list: {student_names}")

# 3) Find a student
print("\\n[2] Find a Student")
search_name = input("Enter a name to search for: ")

if search_name in student_names:
    position = student_names.index(search_name)
    print(f"Found '{search_name}' at index {position}")
else:
    print(f"'{search_name}' not found in the list")

# 4) Remove a student
print("\\n[3] Remove a Student")
remove_name = input("Enter a name to remove: ")

if remove_name in student_names:
    student_names.remove(remove_name)
    print(f"'{remove_name}' has been removed")
else:
    print(f"'{remove_name}' not found, nothing removed")

print(f"Updated list: {student_names}")

# 5) Sort students alphabetically
print("\\n[4] Sort Students")
student_names.sort()
print(f"Sorted list: {student_names}")

# 6) Count students
print("\\n[5] Count Students")
print(f"Total students: {len(student_names)}")
```

# lab4_2.py - Geometric Calculations (Tuples & Immutability)

```python
# Lab 4.2: Geometric Calculations (Tuples & Immutability)

import math

print("--- Lab 4.2: Geometric Calculations ---")

# 1) Define two points using tuples
point1 = (3, 4)
point2 = (6, 8)

# 2) Access the x and y coordinates
print("\\n[1] Access Coordinates")
print(f"Point 1: x = {point1[0]}, y = {point1[1]}")
print(f"Point 2: x = {point2[0]}, y = {point2[1]}")

# 3) Demonstrate tuple immutability
print("\\n[2] Attempt Modification (Immutability Demonstration)")
# Uncomment the next line to see the TypeError:
# point1[0] = 5
print("A tuple is immutable, so its items cannot be changed after creation.")

# 4) Calculate the distance between the points
print("\\n[3] Calculate Distance")
distance = math.sqrt((point2[0] - point1[0]) ** 2 + (point2[1] - point1[1]) ** 2)
print(f"Distance between point1 {point1} and point2 {point2}: {distance}")
```

**Lab 4.1 Challenge: Menu-Driven Student Management**

```python
# Lab 4.1 Challenge

student_names = []

while True:
    print("\\n--- Student Menu ---")
    print("1. Add a student")
    print("2. Remove a student")
    print("3. Find a student")
    print("4. Sort students")
    print("5. Exit")

    choice = input("Choose a menu (1-5): ")

    if choice == "1":
        name = input("Enter student name: ")
        student_names.append(name)
        print(f"Added: {name}")
        print(f"Current list: {student_names}")

    elif choice == "2":
        name = input("Enter the name to remove: ")
        if name in student_names:
            student_names.remove(name)
            print(f"Removed: {name}")
        else:
            print(f"'{name}' was not found.")
        print(f"Current list: {student_names}")

    elif choice == "3":
        name = input("Enter the name to find: ")
        if name in student_names:
            print(f"Found '{name}' at index {student_names.index(name)}")
        else:
            print(f"'{name}' was not found.")

    elif choice == "4":
        student_names.sort()
        print(f"Sorted list: {student_names}")

    elif choice == "5":
        print("Goodbye!")
        break

    else:
        print("Invalid choice. Please enter a number from 1 to 5.")
```

**Lab 4.2 Challenge: Input Coordinates**

```python
# Lab 4.2 Challenge

import math

x1 = float(input("Enter x for point 1: "))
y1 = float(input("Enter y for point 1: "))
x2 = float(input("Enter x for point 2: "))
y2 = float(input("Enter y for point 2: "))

point1 = (x1, y1)
point2 = (x2, y2)

distance = math.sqrt((point2[0] - point1[0]) ** 2 + (point2[1] - point1[1]) ** 2)

print(f"Point 1: {point1}")
print(f"Point 2: {point2}")
print(f"Distance: {distance}")
```

**Activity 4.1: List vs. Tuple — Scenario Matching**

1. Daily temperatures recorded over one week — List. Temperatures may need to be corrected or updated.
2. RGB values for a specific shade — Tuple. The three values describe one fixed color and normally should stay together unchanged.
3. Shopping cart items — List. Items can be added or removed as the customer shops.
4. Days of the week in order — Tuple. The seven days form a fixed sequence that usually does not change.
5. Student assignment scores — List. Scores may be updated or corrected later.
6. GPS coordinates of a fixed landmark — Tuple. The latitude and longitude represent one fixed location.
Main reason: A list is ordered and mutable (can be changed), while a tuple is ordered and immutable (cannot have its items reassigned).

**Activity 4.2: Interactive To-Do List**

```python
# Activity 4.2: Interactive To-Do List

todo_list = []

while True:
    print("\\n--- To-Do List Menu ---")
    print("1. Add a task")
    print("2. View tasks")
    print("3. Mark a task as complete")
    print("4. Exit")

    choice = input("Choose a menu (1-4): ")

    if choice == "1":
        task = input("Enter a task: ")
        todo_list.append(task)
        print("Task added.")

    elif choice == "2":
        if len(todo_list) == 0:
            print("There are no tasks yet.")
        else:
            print(f"You have {len(todo_list)} task(s):")
            for i in range(len(todo_list)):
                print(f"{i + 1}. {todo_list[i]}")

    elif choice == "3":
        if len(todo_list) == 0:
            print("There are no tasks to complete.")
        else:
            for i in range(len(todo_list)):
                print(f"{i + 1}. {todo_list[i]}")

            try:
                task_number = int(input("Enter the task number to mark complete: "))
                if 1 <= task_number <= len(todo_list):
                    completed_task = todo_list.pop(task_number - 1)
                    print(f"Completed and removed: {completed_task}")
                else:
                    print("Invalid task number.")
            except ValueError:
                print("Please enter a valid number.")

    elif choice == "4":
        print("Goodbye!")
        break

    else:
        print("Invalid choice. Please enter a number from 1 to 4.")
```
