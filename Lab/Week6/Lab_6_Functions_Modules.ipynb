## Lab 6.1: Simple Calculator Functions

```python
# ============================================================
# Lab 6.1: Simple Calculator Functions
# ============================================================

def add(a, b):
    """บวกเลขสองจำนวน แล้วคืนค่าผลลัพธ์ a + b"""
    return a + b

def subtract(a, b):
    """ลบเลขสองจำนวน แล้วคืนค่าผลลัพธ์ a - b"""
    return a - b

def multiply(a, b):
    """คูณเลขสองจำนวน แล้วคืนค่าผลลัพธ์ a * b"""
    return a * b

def divide(a, b):
    """หารเลขสองจำนวน (a / b) ถ้า b เป็น 0 จะคืนข้อความ error แทนการหาร"""
    if b == 0:
        return "Error: Division by zero"
    return a / b

# --- Challenge: power() พร้อม default parameter (exponent=2 คือยกกำลังสอง) ---
def power(base, exponent=2):
    """ยกกำลัง base ** exponent ถ้าไม่ส่ง exponent จะใช้ค่า default เป็น 2 (ยกกำลังสอง)"""
    return base ** exponent


def read_number(prompt):
    """อ่านตัวเลขจากผู้ใช้ ถ้าพิมพ์ไม่ใช่ตัวเลขจะถามซ้ำจนกว่าจะถูกต้อง"""
    while True:
        try:
            return float(input(prompt))
        except ValueError:
            print("Invalid number. Please try again.")


while True:
    print("\n=== Simple Calculator ===")
    print("1. Add")
    print("2. Subtract")
    print("3. Multiply")
    print("4. Divide")
    print("5. Power (Challenge)")
    print("6. Exit")

    choice = input("Enter choice (1-6): ").strip()

    if choice == '6':
        print("Goodbye!")
        break

    if choice not in ('1', '2', '3', '4', '5'):
        print("Invalid choice! Please enter a number from 1 to 6.")
        continue

    if choice == '5':
        base = read_number("Enter base: ")
        exp_text = input("Enter exponent (press Enter for default = 2): ").strip()
        if exp_text == "":
            result = power(base)                 # ใช้ค่า default exponent=2
        else:
            try:
                result = power(base, float(exp_text))
            except ValueError:
                print("Invalid exponent.")
                continue
    else:
        a = read_number("Enter first number: ")
        b = read_number("Enter second number: ")
        if choice == '1':
            result = add(a, b)
        elif choice == '2':
            result = subtract(a, b)
        elif choice == '3':
            result = multiply(a, b)
        else:
            result = divide(a, b)

    print(f"Result: {result}")
```

```python
%%writefile my_utils.py
# my_utils.py  -- Custom module สำหรับ Lab 6.2

def greet(name):
    """แสดงข้อความทักทายผู้ใช้"""
    print(f"Hello, {name}!")

def is_prime(number):
    """คืนค่า True ถ้า number เป็นจำนวนเฉพาะ ไม่เช่นนั้นคืน False"""
    if number < 2:
        return False
    i = 2
    while i * i <= number:       # ตรวจถึงแค่ sqrt(number) ก็พอ
        if number % i == 0:
            return False
        i += 1
    return True
```
